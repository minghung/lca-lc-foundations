```python
from langchain.agents import create_agent

# agent = create_agent(model=model)
# agent = create_agent(model="claude-sonnet-4-6")
agent = create_agent(model="gpt-5-nano")
```

```python
from langchain.messages import HumanMessage, AIMessage

response = agent.invoke({"messages":[HumanMessage(content="What is the capital of New Jersey?")]})
pprint(response)
```

```python
response["messages"][-1].content
```

```python
response = agent.invoke(
    {"messages": [HumanMessage(content="What's the capital of the Moon?"),
    AIMessage(content="The capital of the Moon is Luna City."),
    HumanMessage(content="Interesting, tell me more about Luna City")]}
)

pprint(response)
```

```python
msg = HumanMessage(content="What should I visit in Tokyo if I only have 3 days")

for token, metadata in agent.stream({"messages": [msg]}, stream_mode="messages"):
    if token.content:
        print(token.content, end="", flush=True)
```

#### Structured Prompts

```python
system_prompt = """
You are a science fiction writer, create a space capital city at the user's request.

Please keep to the below structure.

Name: The name of the capital city

Location: where it is based

Vibe: 2-3 words to describe its vibe

Economy: Main industries
"""
scifi_agent = create_agent(model="gpt-5-nano", system_prompt=system_prompt)
response = scifi_agent.invoke({"messages": [question]})
print(response["messages"][1].content)
```

```python
from pydantic import BaseModel

class CapitalCity(BaseModel):
    name: str
    location: str
    vibe: str
    economy: str

agent = create_agent(model="gpt-5-nano", system_prompt="You are a science fiction writer, create a space capital city at the user's request.", response_format=CapitalCity)

question = HumanMessage(content="What's the capital of the Moon?")
response = agent.invoke({"messages": [question]})

print(response["structured_response"])

```
