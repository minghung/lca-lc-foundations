```python
from langchain.agents import create_agent
from langchain.messages import HumanMessage
from pprint import pprint
```

```python
agent = create_agent(model="gpt-5-nano")
question = HumanMessage(content="Hello my name is Seán and my favourite colour is green")
response = agent.invoke({"messages": [question]})
pprint(response)
```

```python
question = HumanMessage(content="What's my favorite color?")
response = agent.invoke({"messages": [question]})
pprint(response["messages"][-1].content)
```

```python
from langgraph.checkpoint.memory import InMemorySaver

agent = create_agent(model="gpt-5-nano", checkpointer=InMemorySaver())
question = HumanMessage(content="Hello my name is Seán and my favourite colour is green")
config = {"configurable": {"thread_id": "1"}}

response = agent.invoke({"messages": [question]}, config=config)
pprint(response)
```

```python
question = HumanMessage(content="What is my favorite colour?")
response = agent.invoke({"messages": [question]}, config=config)
pprint(response["messages"][-1].content)
```

```python
from ipywidgets import FileUpload
from IPython.display import display

uploader = FileUpload(accept=".png", multiple=False)
display(uploader)
```

```python
print(uploader.value)
```

```python
import base64

uploaded_file = uploader.value[0]
content_mv = uploaded_file["content"]
img_bytes = bytes(content_mv)
img_b64 = base64.b64encode(img_bytes).decode("utf-8")
```

```python
multimodal_question = HumanMessage(content=[
    {"type": "text", "text": "Tell me about this capital"},
    {"type": "image", "base64": img_b64, "mime_type": "image/png"}
])

response = agent.invoke(
    {"messages": [multimodal_question]}
)

print(response['messages'][-1].content)
```

```python
import sounddevice as sd
from scipy.io.wavfile import write
import base64
import io
import time
from tqdm import tqdm

# Recording settings
duration = 5  # seconds
sample_rate = 44100

print("Recording...")
audio = sd.rec(int(duration * sample_rate), samplerate=sample_rate, channels=1)
# Progress bar for the duration
for _ in tqdm(range(duration * 10)):   # update 10× per second
    time.sleep(0.1)
sd.wait()
print("Done.")

# Write WAV to an in-memory buffer
buf = io.BytesIO()
write(buf, sample_rate, audio)
wav_bytes = buf.getvalue()

aud_b64 = base64.b64encode(wav_bytes).decode("utf-8")
```
